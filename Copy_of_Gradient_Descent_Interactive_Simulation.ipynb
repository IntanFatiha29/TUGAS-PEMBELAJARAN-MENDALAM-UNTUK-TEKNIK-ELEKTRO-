```python
from IPython.display import display
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import FloatSlider, Button, HBox, VBox, Output
```

```python
#Generate data
np.random.seed(100)
X = np.linspace(0, 4, 20)
y = 20 - 1.5*X + np.random.randn(20)*1.5
```

```python
#Prepare figure area
out = Output()
def plot(w, b):
    with out:
        out.clear_output(wait=True)
        # scatter
        plt.scatter(X, y, c='C0')
        # line
        xx = np.array([0,4])
        plt.plot(xx, w*xx + b, 'C2')
        # residuals
        for xi, yi in zip(X,y):
            plt.vlines(xi, yi, w*xi+b, colors='C3', alpha=0.6)
        plt.ylim(0, 25)
        plt.xlabel('x'); plt.ylabel('y')
        plt.title(f'Loss = {(np.square(y - (w*X+b))).mean():.2f}')
        plt.show()
```

```python
# Widgets
w_slider = FloatSlider(min=-4, max=4, step=0.01, value= -1.43, description='Weight')
b_slider = FloatSlider(min=0, max=24, step=0.1, value=19.48, description='Bias')
lr_slider = FloatSlider(min=1e-8, max=1, step=1e-5, value=1e-5, readout_format='.0e', description='LR')

btn_step = Button(description='Step', button_style='info')
btn_reset = Button(description='Reset', button_style='warning')
```

```python
# Gradient-descent logic
def step(_):
    w, b, lr = w_slider.value, b_slider.value, lr_slider.value
    # compute gradients
    y_pred = w*X + b
    dw = (-2 * (y - y_pred) * X).mean()
    db = (-2 * (y - y_pred)).mean()
    # update
    w_slider.value = w - lr*dw
    b_slider.value = b - lr*db
btn_step.on_click(step)

def reset(_):
    w_slider.value = -1.43
    b_slider.value = 19.48
btn_reset.on_click(reset)
```

```python
# Wire everything together
def update_plot(*args):
    plot(w_slider.value, b_slider.value)

w_slider.observe(update_plot, 'value')
b_slider.observe(update_plot, 'value')
```

```python
# Display
controls = VBox([
    HBox([w_slider, b_slider]),
    HBox([lr_slider, btn_step, btn_reset])
])
display(controls, out)
# Initial draw
plot(w_slider.value, b_slider.value)
```
